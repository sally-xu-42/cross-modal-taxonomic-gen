# Load and Inspect Experiment Results

```python
import os
import json
import pandas as pd
from pathlib import Path
```

```python
RUNS_DIR = Path("/share/data/speech/txu/vlm_semantics/runs")
EVAL_FILENAMES = ["eval_leaf_accuracy.csv", "eval_unseen_accuracy.csv"]
```

```python
ignore_suffixes = ["random"]
bad_evals = set()
rows = []

for run_dir in RUNS_DIR.iterdir():
    if not run_dir.is_dir():
        continue
    if not any(run_dir.name.endswith(s) for s in ignore_suffixes):
        continue

    run_name = run_dir.name

    # mark run as bad if ANY expected eval file is missing
    for eval_filename in EVAL_FILENAMES:
        if not (run_dir / eval_filename).exists():
            bad_evals.add(run_name)

    # load leaf eval if present
    eval_file = run_dir / "eval_leaf_accuracy.csv"
    if eval_file.exists():
        tmp = pd.read_csv(eval_file)
        tmp.insert(0, "run_name", run_name)  # adds run_name as first column
        rows.append(tmp)

df = pd.concat(rows, ignore_index=True) if rows else pd.DataFrame()

print(len(bad_evals))
print(len(df))
```

```python
df.to_csv("/share/data/speech/txu/vlm_semantics/src/plotting/random_leaf_evals.csv", index=False)
```
